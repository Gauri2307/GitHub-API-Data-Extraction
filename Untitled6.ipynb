```python

import requests
import pandas as pd
import time
```

```python

repositories = [
    "pandas-dev/pandas",
    "numpy/numpy",
    "psf/requests",
    "pallets/flask",
    "django/django"
]

headers = {
    "Accept": "application/vnd.github+json",
    "User-Agent": "Python-API-Learning"
}
```

```python

all_data = []

for repo in repositories:

    url = f"https://api.github.com/repos/{repo}"

    try:
        response = requests.get(
            url,
            headers=headers,
            timeout=15
        )

        response.raise_for_status()
        data = response.json()

        all_data.append({
            "Repository": data["full_name"],
            "Stars": data["stargazers_count"],
            "Forks": data["forks_count"],
            "Language": data["language"],
            "Open Issues": data["open_issues_count"],
            "URL": data["html_url"]
        })

        print("Extracted:", repo)

        time.sleep(1)

    except requests.exceptions.RequestException as error:
        print("Error extracting", repo, ":", error)
```

```python

df = pd.DataFrame(all_data)

display(df)

print("Total repositories:", len(df))
```

```python

# Find repository with highest stars
top_repo = df.loc[df["Stars"].idxmax()]

print("Most Starred Repository:")
print(top_repo["Repository"])

print("Total Stars:", top_repo["Stars"])

# Calculate average stars
print("Average Stars:", df["Stars"].mean())
```

```python

# Save to CSV
df.to_csv("github_repositories.csv", index=False)

# Save to Excel
df.to_excel("github_repositories.xlsx", index=False)

print("Files saved successfully!")
```
